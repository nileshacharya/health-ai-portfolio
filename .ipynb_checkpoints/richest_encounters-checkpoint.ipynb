````python
#Find richest encounters
# This function scores each encounter by how many conditions, procedures, and medications are associated with it. 
# List the top 20 encounters that will be the evaluation set.


# For each patient and encounter count, conditions, procedures, and medications

import os
import pandas as pd
import json
from pathlib import Path
from datetime import datetime
#from collections import Counter

# ============================================================
# REPORT WRITER
# ============================================================

class MarkdownReport:
    """
    Simple reusable Markdown report generator.
    Different functions can append observations to the same report.
    """

    def __init__(self, output_file):
        self.output_file = output_file
        self.lines = []

    # --------------------------------------------------------
    # Basic Writing Helpers
    # --------------------------------------------------------

    def add_header(self, text, level=1):
        self.lines.append(f"{'#' * level} {text}\n")

    def add_text(self, text):
        self.lines.append(f"{text}\n")

    def add_bullet_list(self, items):
        for item in items:
            self.lines.append(f"- {item}")
        self.lines.append("")

    def add_numbered_list(self, items):
        for idx, item in enumerate(items, start=1):
            self.lines.append(f"{idx}. {item}")
        self.lines.append("")

    def add_table_from_dataframe(self, df):
        """
        Requires:
            pip install tabulate
        """
        markdown_table = df.to_markdown(index=False)
        self.lines.append(markdown_table)
        self.lines.append("")

    def add_code_block(self, code, language="python"):
        self.lines.append(f"```{language}")
        self.lines.append(code)
        self.lines.append("```")
        self.lines.append("")

    def save(self):
        with open(self.output_file, "w", encoding="utf-8") as f:
            f.write("\n".join(self.lines))

        print(f"Markdown report generated: {self.output_file}")


#load data
def load_data():
    print(os.getcwd())
    data_path = Path("./data/synthea/output/fhir")
    conditions = load_ndjson(data_path / "Conditions.ndjson")
    procedures = load_ndjson(data_path / "Procedures.ndjson")
    medications = load_ndjson(data_path / "Medications.ndjson")
    encounters = load_ndjson(data_path / "Encounters.ndjson")

#    all_resources = conditions + procedures + medications + encounters

    return conditions, procedures, medications, encounters    

# ============================================================
# ANALYSIS FUNCTIONS
# ============================================================

def write_report_metadata(report):
    report.add_header("Healthcare Dataset Analysis Report", level=1)

    report.add_text(
        f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}"
    )

    report.add_text(
        "This report demonstrates how multiple functions can contribute "
        "observations into a single Markdown output file."
    )

def load_ndjson(filepath):
    records = []
    with open(filepath) as f:
        for line in f:
            records.append(json.loads(line))
    return records



def analyze_resource_counts(report, conditions, procedures, medications, encounters):
    """
    Example:
    Count FHIR resource types
    """
    report.add_header("Resource Type Counts", level=2)
    report.add_text(
        "Count of each resource type found."
    )
 
    report.add_text(
        f"Conditions: {len(conditions)}"
    )
    report.add_text(
        f"Procedures: {len(procedures)}"
    )
    report.add_text(
        f"Medications: {len(medications)}"
    )
    report.add_text(
        f"Encounters: {len(encounters)}"
    )

#def richest_encounters_by_amount()
# rank encounters by total_claim_cost
# return top 20 encounters
# get patient details
# print them under a new heading in the analysis report

#def richest_encounters_by_conditions()
# Find the richest encounters by number of Conditions
# 


def main():

    
    #Load data
    conditions, procedures, medications, encounters = load_data()


    # Create report object
    report = MarkdownReport("analysis_report.md")

    # report header
    write_report_metadata(report)

    # report counts
    analyze_resource_counts(report, conditions, procedures, medications, encounters)

    
    # Find the richest encounters by $Amount
    #richest_encounters_by_amount()

    # Find the richest encounters by number of Conditions

    # Find the richest encounters by number of Medications

    # Find the richest encounters by number of Procedures

    # Find patient details of a given encounter



    

    # Save markdown file
    report.save()


if __name__ == "__main__":
    main()
````

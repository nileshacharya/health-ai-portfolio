````python
#import os
#from mdutils.mdutils import MdUtils

# Create a Markdown file
#md_file = MdUtils(file_name='example', title='My Markdown File')
#md_file.new_header(level=1, title='Atx Header 1')
#md_file.new_paragraph("This is a paragraph in **Markdown**.")

# Save file
#md_file.create_md_file()

"""
example_markdown_report.py

Purpose:
--------
Demonstrates how a data science / analytics script can generate
a single Markdown report from observations collected across
multiple functions.

This pattern is useful for:
- Data profiling
- Healthcare/FHIR analysis
- Synthea output analysis
- ETL validation
- Audit reporting
- ML experiment summaries

Output:
-------
analysis_report.md
"""

from datetime import datetime
from collections import Counter
import pandas as pd


# ============================================================
# REPORT WRITER
# ============================================================

class MarkdownReport:
    """
    Simple reusable Markdown report generator.
    Different functions can append observations to the same report.
    """

    def __init__(self, output_file):
        self.output_file = output_file
        self.lines = []

    # --------------------------------------------------------
    # Basic Writing Helpers
    # --------------------------------------------------------

    def add_header(self, text, level=1):
        self.lines.append(f"{'#' * level} {text}\n")

    def add_text(self, text):
        self.lines.append(f"{text}\n")

    def add_bullet_list(self, items):
        for item in items:
            self.lines.append(f"- {item}")
        self.lines.append("")

    def add_numbered_list(self, items):
        for idx, item in enumerate(items, start=1):
            self.lines.append(f"{idx}. {item}")
        self.lines.append("")

    def add_table_from_dataframe(self, df):
        """
        Requires:
            pip install tabulate
        """
        markdown_table = df.to_markdown(index=False)
        self.lines.append(markdown_table)
        self.lines.append("")

    def add_code_block(self, code, language="python"):
        self.lines.append(f"```{language}")
        self.lines.append(code)
        self.lines.append("```")
        self.lines.append("")

    def save(self):
        with open(self.output_file, "w", encoding="utf-8") as f:
            f.write("\n".join(self.lines))

        print(f"Markdown report generated: {self.output_file}")


# ============================================================
# SAMPLE DATA
# ============================================================

def load_sample_data():
    """
    Simulated healthcare/FHIR-style dataset.
    """

    patients = [
        {"patient_id": 1, "condition": "Hypertension", "resource_type": "Condition"},
        {"patient_id": 2, "condition": "Diabetes", "resource_type": "Condition"},
        {"patient_id": 3, "condition": "Hypertension", "resource_type": "Condition"},
        {"patient_id": 4, "condition": "Asthma", "resource_type": "Condition"},
        {"patient_id": 5, "condition": "Diabetes", "resource_type": "Condition"},
        {"patient_id": 6, "condition": "Diabetes", "resource_type": "Condition"},
        {"patient_id": 7, "condition": "Hypertension", "resource_type": "Condition"},
    ]

    observations = [
        {"resource_type": "Observation"},
        {"resource_type": "Observation"},
        {"resource_type": "Observation"},
    ]

    medications = [
        {"resource_type": "MedicationRequest"},
        {"resource_type": "MedicationRequest"},
    ]

    all_resources = patients + observations + medications

    return patients, all_resources


# ============================================================
# ANALYSIS FUNCTIONS
# ============================================================

def write_report_metadata(report):
    report.add_header("Healthcare Dataset Analysis Report", level=1)

    report.add_text(
        f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}"
    )

    report.add_text(
        "This report demonstrates how multiple functions can contribute "
        "observations into a single Markdown output file."
    )


def analyze_resource_counts(report, resources):
    """
    Example:
    Count FHIR resource types
    """

    report.add_header("Resource Type Counts", level=2)

    resource_counts = Counter(
        resource["resource_type"]
        for resource in resources
    )

    df = pd.DataFrame(
        [
            {"Resource Type": key, "Count": value}
            for key, value in resource_counts.items()
        ]
    )

    report.add_text(
        "The following table shows the count of each resource type found."
    )

    report.add_table_from_dataframe(df)

    # Add observations
    report.add_header("Observations", level=3)

    observations = [
        f"Total distinct resource types: {len(resource_counts)}",
        f"Most common resource type: {max(resource_counts, key=resource_counts.get)}",
    ]

    report.add_bullet_list(observations)


def analyze_top_conditions(report, patients):
    """
    Example:
    Top conditions analysis
    """

    report.add_header("Top Conditions", level=2)

    condition_counts = Counter(
        patient["condition"]
        for patient in patients
    )

    top_conditions = condition_counts.most_common(3)

    df = pd.DataFrame(
        top_conditions,
        columns=["Condition", "Count"]
    )

    report.add_text(
        "Top 3 conditions identified in the dataset."
    )

    report.add_table_from_dataframe(df)

    report.add_header("Key Findings", level=3)

    findings = [
        "Hypertension and Diabetes are the dominant conditions.",
        "Condition distribution may indicate chronic disease-heavy population.",
        "Further demographic analysis is recommended.",
    ]

    report.add_numbered_list(findings)


def write_recommendations(report):
    """
    Example:
    Plain narrative text + bullets
    """

    report.add_header("Recommendations", level=2)

    report.add_text(
        "Based on the analysis performed above, the following next steps "
        "are recommended for the engineering and analytics teams."
    )

    recommendations = [
        "Validate resource completeness across all exports.",
        "Introduce automated anomaly detection for missing data.",
        "Track longitudinal condition trends over time.",
        "Add demographic segmentation analysis.",
    ]

    report.add_bullet_list(recommendations)


def write_sample_query(report):
    """
    Example:
    Add code block into markdown
    """

    report.add_header("Example Analysis Query", level=2)

    sample_code = """
top_conditions = (
    df["condition"]
    .value_counts()
    .head(3)
)
"""

    report.add_text(
        "Below is an example pandas query used during analysis."
    )

    report.add_code_block(sample_code, language="python")


# ============================================================
# MAIN
# ============================================================

def main():

    # Load data
    patients, resources = load_sample_data()

    # Create report object
    report = MarkdownReport("analysis_report.md")

    # Multiple functions contribute to SAME markdown file
    write_report_metadata(report)
    analyze_resource_counts(report, resources)
    analyze_top_conditions(report, patients)
    write_recommendations(report)
    write_sample_query(report)

    # Save markdown file
    report.save()


if __name__ == "__main__":
    main()

````

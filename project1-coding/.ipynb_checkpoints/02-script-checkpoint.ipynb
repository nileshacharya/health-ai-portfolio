```python
###
### Query Synthea's actual codes for this encounter
###
###

import psycopg2
import pandas as pd
import json

conn = psycopg2.connect(dbname="health_ai_portfolio", host="localhost")

def query(sql, params=None):
    return pd.read_sql(sql, conn, params=params)


def query(sql, params=None):
    return pd.read_sql(sql, conn, params=params)

df = query("""
    SELECT
        c.code                  AS icd10_code,
        c.display               AS icd10_display,
        c.clinical_status
    FROM conditions c
    WHERE c.encounter_id = '9917e9f0-753a-4673-da3f-ca0de161db03'
    ORDER BY c.recorded_date
""")
print("=== GROUND TRUTH ICD-10 ===")
print(df.to_string(index=False))

df2 = query("""
    SELECT
        p.code                  AS procedure_code,
        p.display               AS procedure_display,
        p.status
    FROM procedures p
    WHERE p.encounter_id = '9917e9f0-753a-4673-da3f-ca0de161db03'
    GROUP BY p.code, p.display, p.status
    ORDER BY p.code
""")
print("\n=== GROUND TRUTH PROCEDURES ===")
print(df2.to_string(index=False))
```

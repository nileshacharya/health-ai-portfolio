```python
"""
Query 3: Richness of data:
Richness Scoring
-- conditions weighted 3x, 
-- medications 2x, procedures capped at 20,
-- observations as supporting signal only,
-- zero conditions = penalized
"""
df = query("""
    WITH encounter_richness AS (
        SELECT
            e.id                                AS encounter_id,
            e.patient_id,
            e.encounter_class,
            e.encounter_type,
            e.start_date::DATE                  AS encounter_date,
            COUNT(DISTINCT c.id)                AS condition_count,
            COUNT(DISTINCT p.id)                AS procedure_count,
            COUNT(DISTINCT m.id)                AS medication_count,
            COUNT(DISTINCT o.id)                AS observation_count,
            CASE 
                WHEN COUNT(DISTINCT c.id) = 0 THEN 0
                ELSE
                    COUNT(DISTINCT c.id) * 3
                    + LEAST(COUNT(DISTINCT p.id), 20) * 2
                    + COUNT(DISTINCT m.id) * 2
                    + LEAST(COUNT(DISTINCT o.id), 50)
            END                                 AS richness_score
        FROM encounters e
        LEFT JOIN conditions           c ON c.encounter_id = e.id
        LEFT JOIN procedures           p ON p.encounter_id = e.id
        LEFT JOIN medication_requests  m ON m.encounter_id = e.id
        LEFT JOIN observations         o ON o.encounter_id = e.id
        GROUP BY
            e.id, e.patient_id, e.encounter_class,
            e.encounter_type, e.start_date
    )
    SELECT *,
        RANK() OVER (ORDER BY richness_score DESC) AS richness_rank
    FROM encounter_richness
    WHERE richness_score > 0
    ORDER BY richness_score DESC
    LIMIT 25
""")
df
```

```python
"""
Query 4: Musculoskeletal and rehabilitation encounters:
"""

df = query("""
    SELECT
        X.encounter_id,
        X.encounter_type,
        X.encounter_date,
        X.richness_score,
        STRING_AGG(c.code || ' · ' || c.display, E'\n'
                   ORDER BY c.code)     AS icd10_codes
    FROM (
        WITH er AS (
            SELECT
                e.id            AS encounter_id,
                e.patient_id,
                e.encounter_type,
                e.start_date::DATE AS encounter_date,
                COUNT(DISTINCT c.id)
                    + COUNT(DISTINCT p.id) * 2
                    + COUNT(DISTINCT m.id) * 2  AS richness_score
            FROM encounters e
            LEFT JOIN conditions          c ON c.encounter_id = e.id
            LEFT JOIN procedures          p ON p.encounter_id = e.id
            LEFT JOIN medication_requests m ON m.encounter_id = e.id
            GROUP BY e.id, e.patient_id, e.encounter_type, e.start_date
        ) SELECT * FROM er
    ) X
    JOIN conditions c ON c.encounter_id = X.encounter_id
    WHERE
        c.code LIKE 'M%'
        OR c.code LIKE 'Z47%'
        OR c.code LIKE 'Z96%'
        OR c.code LIKE 'Z87.3%'
        OR X.encounter_type ILIKE '%physical%'
        OR X.encounter_type ILIKE '%orthop%'
        OR X.encounter_type ILIKE '%rehab%'
    GROUP BY
        X.encounter_id, X.encounter_type,
        X.encounter_date, X.richness_score
    ORDER BY X.richness_score DESC
    LIMIT 20
""")
df
```

```python
"""
Query 5:  Insurance data preview
"""

df = query("""
    SELECT
        e.encounter_type,
        e.encounter_class,
        COUNT(*)                             AS claim_count,
        ROUND(AVG(eob.total_cost), 2)        AS avg_total_cost,
        ROUND(MIN(eob.total_cost), 2)        AS min_total_cost,
        ROUND(MAX(eob.total_cost), 2)        AS max_total_cost,
        ROUND(AVG(eob.insurance_paid), 2)    AS avg_insurance_paid,
        ROUND(AVG(
            eob.total_cost - COALESCE(eob.insurance_paid, 0)
        ), 2)                                AS avg_patient_responsibility,
        ROUND(
            AVG(eob.insurance_paid)
            / NULLIF(AVG(eob.total_cost), 0) * 100
        , 1)                                 AS pct_covered
    FROM explanation_of_benefits eob
    JOIN encounters e ON e.id = eob.encounter_id
    WHERE eob.total_cost > 0
    GROUP BY e.encounter_type, e.encounter_class
    ORDER BY avg_total_cost DESC
    LIMIT 20
""")
df
```

```python
"""
Query 5a:  Insurance data preview - highest avg_total_cost claims ordered by pct_covered 
"""

df = query("""
    Select * from 
    (SELECT
        e.encounter_type,
        e.encounter_class,
        COUNT(*)                             AS claim_count,
        ROUND(AVG(eob.total_cost), 2)        AS avg_total_cost,
        ROUND(MIN(eob.total_cost), 2)        AS min_total_cost,
        ROUND(MAX(eob.total_cost), 2)        AS max_total_cost,
        ROUND(AVG(eob.insurance_paid), 2)    AS avg_insurance_paid,
        ROUND(AVG(
            eob.total_cost - COALESCE(eob.insurance_paid, 0)
        ), 2)                                AS avg_patient_responsibility,
        ROUND(
            AVG(eob.insurance_paid)
            / NULLIF(AVG(eob.total_cost), 0) * 100
        , 1)                                 AS pct_covered
    FROM explanation_of_benefits eob
    JOIN encounters e ON e.id = eob.encounter_id
    WHERE eob.total_cost > 0
    GROUP BY e.encounter_type, e.encounter_class
    ORDER BY avg_total_cost DESC
    LIMIT 20) x
    order by x.pct_covered
""")
df
```

```python
"""
Query 5b — Per-encounter cost rollup instead of per claim (per explanation of benefit)
One clinical encounter generates multiple billing records.
Analyze cost per encounter — summing all EOB line items per encounter rather than averaging individual EOBs. 

"""
df = query("""
    SELECT
        e.encounter_type,
        COUNT(DISTINCT e.id)                        AS encounter_count,
        ROUND(AVG(encounter_totals.total), 2)        AS avg_cost_per_encounter,
        ROUND(AVG(encounter_totals.paid), 2)         AS avg_paid_per_encounter,
        ROUND(AVG(encounter_totals.total 
              - COALESCE(encounter_totals.paid, 0)), 2) AS avg_gap_per_encounter,   
        ROUND(AVG(encounter_totals.total - COALESCE(encounter_totals.paid, 0)) /
            AVG(encounter_totals.total) * 100, 1)        
    FROM encounters e
    JOIN (
        SELECT
            encounter_id,
            SUM(total_cost)     AS total,
            SUM(insurance_paid) AS paid
        FROM explanation_of_benefits
        WHERE total_cost > 0
        GROUP BY encounter_id
    ) encounter_totals ON encounter_totals.encounter_id = e.id
    GROUP BY e.encounter_type
    ORDER BY avg_cost_per_encounter DESC
    LIMIT 15
""")
df
```

## Query 5B Observations ##
The per-encounter rollup tells a materially different story than the per-EOB average. 
**Hospital admission** jumps from $6,683 (Query 5) to $18,517 (Query 5B) once you sum all line items per encounter. 
A single inpatient stay generates a facility fee, a physician fee, lab fees, and imaging fees as separate claims. 
Always use the rollup for cost analysis, never raw EOB averages.
**Orthopedic** At $15,675 average per encounter with a $5,774 gap — meaning insurers leave $5,774 uncovered per orthopedic admission — this is a high-stakes, high-friction encounter type where prior auth AI creates direct measurable value. The gap is large enough to matter clinically and financially, and the encounter count (16) is representative without being an edge case.
**Surgical admissions** (row 12) are the most interesting anomaly. $6,799 average cost but only $1,085 paid — an 84% gap. 
Combined with row 14 (surgical transplant) at $3,421 gap, surgical encounters are systematically underpaid relative to cost. 
This pattern in real data would trigger a revenue cycle investigation. 
Identifying clinically meaningful patterns like this is very important - SQL skills are important

```python
"""
Query 5c — Insurer breakdown aka payer mix in the dataset.
Reveals relative coverage rates of different payers in the dataset — Medicare, Medicaid, commercial etc. 
Payer mix determines reimbursement strategy, prior auth requirements, and product go-to-market motion. 
Different payers have different prior auth rules and different API mandates under the CMS interoperability rule.

"""
df = query("""
    SELECT
        resource -> 'insurer' ->> 'display'     AS insurer,
        COUNT(*)                                AS claim_count,
        ROUND(AVG(total_cost), 2)               AS avg_total_cost,
        ROUND(AVG(insurance_paid), 2)           AS avg_paid,
        ROUND(AVG(insurance_paid) 
              / NULLIF(AVG(total_cost), 0) * 100
        , 1)                                    AS pct_covered
    FROM explanation_of_benefits
    WHERE total_cost > 0
    GROUP BY resource -> 'insurer' ->> 'display'
    ORDER BY claim_count DESC
""")
df
```

# Readme #
** Decision 1 — Richness scoring criteria.** Conditions weighted 3x because they drive ICD-10 coding. Procedures capped at 20 to prevent workflow events from dominating. Zero-condition encounters excluded from the coding evaluation set.
** Decision 2 — Evaluation set composition. ** 12 encounters across 4 encounter types, 2 care settings, 3 clinical clusters. Deliberately includes social determinants, post-surgical follow-up, and acute respiratory illness to test coding breadth.
** Decision 3 — SNOMED to ICD-10 translation delegated to the AI. ** Rather than pre-mapping SNOMED codes to ICD-10 in SQL, the context assembly passes clinical descriptions and lets Claude suggest the appropriate ICD-10 codes. This is a product decision with a rationale — clinicians document in clinical language, coders translate to billing language, and the AI bridges that gap.
